```python
import pandas as pd
import joblib
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.multioutput import MultiOutputClassifier
from config import model_type, features, target, test_size, random_state, n_estimators, max_depth

data = pd.read_csv("training_data.csv")

X = data[features]
y = data[target] if isinstance(target, str) else data[target]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=test_size, random_state=random_state)

if model_type == "single_output":
    model = RandomForestClassifier(n_estimators=n_estimators, max_depth=max_depth, random_state=random_state)
else:
    model = MultiOutputClassifier(RandomForestClassifier(n_estimators=n_estimators, max_depth=max_depth, random_state=random_state))

model.fit(X_train, y_train)
print(f"Accuracy: {model.score(X_test, y_test)*100:.2f}%")

joblib.dump(model, "model.pkl")
print("[OK] model.pkl saved")
```
